```python
# closers basically allows us to access the variable of the outer funtion
```

```python
# outer funtion
def outer(name):
    # inner funtion
    def inner():
        print("Hello", name)
        
    inner()    
outer("ankit")
```

```python
def num_square(num):
    def square():
        print("square is", num ** 2)

    print("number is: ", num)
    square()

for num in range(1, 10):
    num_square(num)

```

```python
def calculate(n):
    def multipler(x):
        return x * n
    return multipler

mult = calculate(3)
print(mult(5))
```

```python
def calculate(n):
    def multipler(x):
        return n
    return multipler

mult = calculate(3)
print(mult.__closure__)
```

```python
def calculate(n):
    def multipler(x):
        return n
    return multipler

mult = calculate(3)
print(mult.__closure__[0].cell_contents)
```
